```python
import pandas as pd
```

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv')
```

```python
# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year',
                       'genre', 'country']]

df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

# merge()

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician',
                            'Accountant', 'Developer']})
```

```python
df1
```

```python
df2
```

## Inner join

```python
df1.merge(df2, on='id', how='inner')
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='inner').shape)
```

### Exercise

```python
# merge df_movies and df_ratings (inner join)
df_movies.merge(df_ratings, on='imdb_title_id')
# pd.merge(df_movies, df_ratings, on='imdb_title_id')
```

## Outer join (Full join)

```python
df1.merge(df2, on='id', how='outer')
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='outer').shape)
```

### Exercise

```python
# merge df_movies and df_ratings (outer join)
df_movies.merge(df_ratings, on='imdb_title_id', how='outer')
```

## Exclusive Outer join (Exclusive Full join)

```python
df1.merge(df2, on='id', how='outer')
```

```python
# indicator=True
df1.merge(df2, on='id', how='outer',
          indicator=True)
```

```python
# query
df1.merge(df2, on='id', how='outer',
          indicator=True).query("_merge=='left_only' or _merge=='right_only'")
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='outer',
          indicator=True).query("_merge=='left_only' or _merge=='right_only'").shape)
```

### Exercise

```python
# merge df_movies and df_ratings
df_movies.merge(df_ratings, on='imdb_title_id', how='outer', 
                indicator=True).query("_merge=='left_only' or _merge=='right_only'")
```

##  Left join

```python
df1.merge(df2, on='id', how='left')
```

### Exercise

```python
# extract a 50% sample of the df_movies dataframe
df_movies_sample = df_movies.sample(frac=0.5)
```

```python
# merge df_movies_sample and df_ratings (left join)
df_left = df_movies_sample.merge(df_ratings, on='imdb_title_id',
                                 how='left')
```

```python
# shape
print(df_movies_sample.shape)
print(df_ratings.shape)
print(df_left.shape)
```

## Exclusive Left join

```python
# indicator=True
df1.merge(df2, on='id', how='outer',
          indicator = True)
```

```python
# query
df1.merge(df2, on='id', how='outer',
          indicator = True).query("_merge=='left_only'")
```

```python
# left join
pd.merge(df_movies, df_ratings.sample(frac=.5),
         on='imdb_title_id', how='left').shape
```

### Exercise

```python
# make a copy of the df_movies dataframe
df_movies_2 = df_movies.copy()
```

```python
df_movies_2.head()
```

```python
# set the first 1000 values of 'imdb_title_id' column as 'tt1234567890'
for index in df_movies_2.index:
    if index < 1000:
        df_movies_2.loc[index, 'imdb_title_id'] = 'tt1234567890'
```

```python
df_movies_2.head()
```

```python
# merge df_movies_2 and df_ratings (exclusive left join)
df_exclusive_left = df_movies_2.merge(df_ratings,
                                      on='imdb_title_id',
                                      how='outer', 
                                      indicator=True).query("_merge=='left_only'")
```

```python
df_exclusive_left.head()
```

```python
# shape
print(df_movies_2.shape)
print(df_ratings.shape)
print(df_exclusive_left.shape)
```
